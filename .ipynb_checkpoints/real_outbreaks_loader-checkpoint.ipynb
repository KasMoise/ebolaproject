# Real outbreak loaders (West Africa 2014-16, Uganda 2022)
1. Run the first code cell (defines all functions).
2. Run the second cell (downloads the files and prints their columns).
3. To use this in your main notebook, copy the first code cell into it.

```python
"""
real_outbreaks.py - loaders for REAL historical Ebola outbreak data
(West Africa 2014-16, Uganda 2022) to replace the simulated SIR sources in
the transfer-learning module.

Everything is harmonised to one format:  zone | date (week start, Monday) | value (NEW cases/week)

Typical use in the notebook
---------------------------
    import real_outbreaks as ro

    # 1) discover + download (needs internet), then look at the columns
    #    python real_outbreaks.py --download

    # 2) load (override auto-detected columns if needed)
    wa  = ro.load_west_africa("real_sources/hdx/ebola_data_db_format.csv",
                              indicator_regex=r"cumulative.*confirmed.*cases")
    ug  = ro.load_uganda_2022("real_sources/github/....csv")

    # 3) build pre-training windows (input_dim = 1, matches EpidemicEncoder)
    corpus = ro.build_corpus({"WestAfrica_2014": wa, "Uganda_2022": ug}, seq_len=4)
    X, y, groups = corpus["X"], corpus["y"], corpus["groups"]

    # 4) put the 2026 target on the SAME (weekly) scale
    target_weekly = ro.counts_to_weekly(DATA["cum_cases"], "date", "zone", "value")

IMPORTANT: sources are WEEKLY. The notebook's 2026 data is DAILY (SEQ_LEN=7 days).
Use weekly for both source and target when transferring.

Column names of the remote files were NOT verified when this was written:
call inspect(df) and pass columns={...} to override the guesses.
"""
from __future__ import annotations

import argparse
import re
import warnings
from pathlib import Path
from typing import Dict, List, Optional, Sequence

import numpy as np
import pandas as pd
import requests

HDX_PACKAGE_API = "https://data.humdata.org/api/3/action/package_show"
GITHUB_CONTENTS_API = "https://api.github.com/repos/{repo}/contents/{path}"
HEADERS = {"User-Agent": "real-outbreaks-loader/1.0"}
TOTAL = "__total__"

SOURCES = {
    "WestAfrica_2014": {"strain": "Zaire", "hdx_package": "ebola-cases-2014"},
    "Uganda_2022": {"strain": "Sudan", "github_repo": "globaldothealth/ebola"},
}


# --------------------------------------------------------------------------
# Download helpers
# --------------------------------------------------------------------------
def _http_get(url: str, timeout: int = 60, **kw) -> requests.Response:
    r = requests.get(url, headers=HEADERS, timeout=timeout, **kw)
    r.raise_for_status()
    return r


def fetch_hdx(package_id: str, dest, formats=("csv", "xlsx", "xls")) -> List[Path]:
    """Download the CSV/Excel resources of an HDX (CKAN) dataset."""
    dest = Path(dest)
    dest.mkdir(parents=True, exist_ok=True)
    meta = _http_get(HDX_PACKAGE_API, params={"id": package_id}).json()
    if not meta.get("success"):
        raise RuntimeError(f"HDX package '{package_id}' not found")
    paths = []
    for res in meta["result"].get("resources", []):
        fmt = (res.get("format") or "").lower()
        url = res.get("download_url") or res.get("url")
        if fmt not in formats or not url:
            continue
        out = dest / (Path(url.split("?")[0]).name or f"{res['id']}.{fmt}")
        if not out.exists():
            out.write_bytes(_http_get(url, timeout=120).content)
        paths.append(out)
        print(f"  HDX: {out.name}")
    return paths


def fetch_github_files(repo: str, dest, path: str = "",
                       suffixes=(".csv", ".csv.gz"), max_depth: int = 3) -> List[Path]:
    """Download CSV files from a public GitHub repo (unauthenticated: 60 API calls/hour)."""
    dest = Path(dest)
    dest.mkdir(parents=True, exist_ok=True)
    found: List[Path] = []

    def walk(p: str, depth: int) -> None:
        items = _http_get(GITHUB_CONTENTS_API.format(repo=repo, path=p)).json()
        if isinstance(items, dict):
            items = [items]
        for it in items:
            if it["type"] == "dir" and depth < max_depth:
                walk(it["path"], depth + 1)
            elif it["type"] == "file" and it["name"].lower().endswith(suffixes) and it.get("download_url"):
                out = dest / it["path"].replace("/", "__")
                if not out.exists():
                    out.write_bytes(_http_get(it["download_url"], timeout=120).content)
                found.append(out)
                print(f"  GitHub: {out.name}")

    walk(path, 0)
    return found


# --------------------------------------------------------------------------
# Reading and column detection
# --------------------------------------------------------------------------
def _norm(s) -> str:
    return re.sub(r"[^a-z0-9]", "", str(s).lower())


def read_table(path) -> pd.DataFrame:
    path = Path(path)
    if path.name.lower().endswith((".xlsx", ".xls")):
        return pd.read_excel(path)
    err = None
    for enc in ("utf-8", "utf-8-sig", "latin-1"):
        try:
            return pd.read_csv(path, encoding=enc, low_memory=False)
        except UnicodeDecodeError as e:
            err = e
    raise err  # type: ignore[misc]


def parse_dates(s: pd.Series) -> pd.Series:
    try:
        out = pd.to_datetime(s, errors="coerce", format="mixed")
    except (TypeError, ValueError):
        out = pd.to_datetime(s, errors="coerce")
    if getattr(out.dt, "tz", None) is not None:
        out = out.dt.tz_localize(None)
    return out


def _is_text(s: pd.Series) -> bool:
    return not (pd.api.types.is_numeric_dtype(s) or pd.api.types.is_datetime64_any_dtype(s))


def guess_columns(df: pd.DataFrame,
                  date_prefer=("onset", "confirm", "report"),
                  zone_prefer=("district", "adminarealevel2", "admin2", "adminarealevel3",
                               "location", "province", "region", "country")) -> Dict[str, Optional[str]]:
    """Heuristic guess of date / zone / status / value / indicator columns."""
    cols = list(df.columns)
    nm = {c: _norm(c) for c in cols}
    out: Dict[str, Optional[str]] = dict(date=None, zone=None, status=None, value=None, indicator=None)

    # date: columns that parse as dates, preferring onset > confirmation > report
    scored = {}
    for c in cols:
        if not any(h in nm[c] for h in ("date",) + tuple(date_prefer)):
            continue
        s = df[c].dropna()
        if s.empty or pd.api.types.is_numeric_dtype(s):
            continue
        ratio = parse_dates(s.astype(str).head(300)).notna().mean()
        if ratio >= 0.6:
            scored[c] = ratio
    for hint in date_prefer:
        hit = [c for c in scored if hint in nm[c]]
        if hit:
            out["date"] = hit[0]
            break
    if out["date"] is None and scored:
        out["date"] = max(scored, key=scored.get)

    # zone
    for hint in zone_prefer:
        for c in cols:
            if hint in nm[c] and c != out["date"] and _is_text(df[c]) and 2 <= df[c].nunique() <= 500:
                out["zone"] = c
                break
        if out["zone"]:
            break

    # status / indicator / value
    for c in cols:
        if out["status"] is None and any(h in nm[c] for h in ("status", "classification", "casedef")) \
                and _is_text(df[c]) and 1 <= df[c].nunique() <= 12:
            out["status"] = c
        if out["indicator"] is None and nm[c] in ("indicator", "variable", "series", "metric"):
            out["indicator"] = c
        if out["value"] is None and pd.api.types.is_numeric_dtype(df[c]) \
                and (nm[c] in ("value", "cases", "count", "n") or nm[c].endswith("cases")):
            out["value"] = c
    return out


def inspect(df: pd.DataFrame, n: int = 3) -> Dict[str, Optional[str]]:
    """Print columns, a few values, and the guessed roles."""
    print(f"{len(df):,} rows x {df.shape[1]} columns")
    for c in df.columns:
        sample = df[c].dropna().astype(str).unique()[:n].tolist()
        print(f"  {str(c)[:45]:45s} {str(df[c].dtype):10s} nunique={df[c].nunique():<6} e.g. {sample}")
    g = guess_columns(df)
    print("guessed roles:", g)
    return g


# --------------------------------------------------------------------------
# Conversion to weekly NEW cases
# --------------------------------------------------------------------------
def _week_start(dates: pd.Series) -> pd.Series:
    return dates.dt.to_period("W-SUN").dt.start_time  # Monday


def linelist_to_weekly(df: pd.DataFrame, date_col: str, zone_col: Optional[str] = None,
                       status_col: Optional[str] = None,
                       keep_status: Optional[Sequence[str]] = None,
                       pad_to_end: bool = True) -> pd.DataFrame:
    """Individual-level records -> weekly new cases per zone."""
    d = df.copy()
    d["_date"] = parse_dates(d[date_col])
    d = d.dropna(subset=["_date"])
    if status_col and keep_status:
        keep = {k.lower() for k in keep_status}
        d = d[d[status_col].astype(str).str.strip().str.lower().isin(keep)]
    if d.empty:
        raise ValueError("No rows left after date/status filtering - check columns and keep_status")
    d["_zone"] = d[zone_col].fillna("Unknown").astype(str).str.strip() if zone_col else "ALL"
    d["_week"] = _week_start(d["_date"])
    wk = d.groupby(["_zone", "_week"]).size().rename("value").reset_index()
    last = wk["_week"].max()
    frames = []
    for z, g in wk.groupby("_zone"):
        g = g.set_index("_week")["value"]
        idx = pd.date_range(g.index.min(), last if pad_to_end else g.index.max(), freq="7D")
        frames.append(pd.DataFrame({"zone": z, "date": idx,
                                    "value": g.reindex(idx, fill_value=0).to_numpy(float)}))
    return pd.concat(frames, ignore_index=True)


def counts_to_weekly(df: pd.DataFrame, date_col: str, zone_col: Optional[str], value_col: str,
                     cumulative: bool = True, verbose: bool = False) -> pd.DataFrame:
    """Aggregate counts (cumulative or incident) -> weekly new cases per zone.

    Cumulative series are forced monotone (reporting corrections are ignored),
    then differenced. The first week holds the cases reported up to that date."""
    d = pd.DataFrame({
        "date": parse_dates(df[date_col]),
        "zone": df[zone_col].astype(str).str.strip() if zone_col else "ALL",
        "value": pd.to_numeric(df[value_col], errors="coerce"),
    }).dropna(subset=["date", "value"])
    d["week"] = _week_start(d["date"])
    wk = d.groupby(["zone", "week"])["value"].agg("max" if cumulative else "sum").reset_index()
    frames, n_corr = [], 0
    for z, g in wk.groupby("zone"):
        g = g.set_index("week")["value"].sort_index()
        s = g.reindex(pd.date_range(g.index.min(), g.index.max(), freq="7D"))
        if cumulative:
            s = s.ffill()
            n_corr += int((s.diff() < 0).sum())
            s = s.cummax()
            new = s.diff()
            new.iloc[0] = s.iloc[0]
        else:
            new = s.fillna(0)
        frames.append(pd.DataFrame({"zone": z, "date": new.index, "value": new.to_numpy(float)}))
    if verbose and n_corr:
        print(f"  {n_corr} downward corrections in cumulative counts were ignored")
    return pd.concat(frames, ignore_index=True)


def add_total(weekly: pd.DataFrame) -> pd.DataFrame:
    """Append a '__total__' series (sum over zones). No-op if there is a single zone."""
    base = weekly[weekly["zone"] != TOTAL]
    if base["zone"].nunique() < 2:
        return base.reset_index(drop=True)
    tot = base.groupby("date", as_index=False)["value"].sum()
    tot["zone"] = TOTAL
    return pd.concat([base, tot[["zone", "date", "value"]]], ignore_index=True)


# --------------------------------------------------------------------------
# Validation
# --------------------------------------------------------------------------
def validate(weekly: pd.DataFrame, name: str, expected_total: Optional[float] = None,
             tol: float = 0.25) -> Dict:
    base = weekly[weekly["zone"] != TOTAL]
    by_week = base.groupby("date")["value"].sum()
    rep = dict(name=name, zones=int(base["zone"].nunique()), weeks=int(by_week.size),
               start=by_week.index.min().date(), end=by_week.index.max().date(),
               total=float(base["value"].sum()), peak_week=by_week.idxmax().date(),
               peak_value=float(by_week.max()))
    print(f"[{name}] {rep['zones']} zones | {rep['weeks']} weeks | {rep['start']} -> {rep['end']} | "
          f"total={rep['total']:,.0f} | peak {rep['peak_value']:,.0f} on {rep['peak_week']}")
    if expected_total and abs(rep["total"] - expected_total) / expected_total > tol:
        warnings.warn(f"{name}: total {rep['total']:,.0f} differs from expected ~{expected_total:,.0f} by more "
                      f"than {tol:.0%}. Wrong indicator / case classification / file? Check before using.")
    if (base["value"] < 0).any():
        warnings.warn(f"{name}: negative values present")
    return rep


# --------------------------------------------------------------------------
# High-level loaders
# --------------------------------------------------------------------------
def load_outbreak(path, name: str, kind: str = "auto", columns: Optional[Dict[str, str]] = None,
                  indicator_regex: Optional[str] = None, cumulative: bool = True,
                  keep_status: Optional[Sequence[str]] = None,
                  expected_total: Optional[float] = None, verbose: bool = True) -> pd.DataFrame:
    """Load one file -> weekly new cases (zone | date | value).

    kind: 'counts' (aggregated, has a numeric value column), 'linelist' (one row per case) or 'auto'.
    columns: override guesses, e.g. {'date': 'DateOnset', 'zone': 'District', 'value': 'value'}.
    indicator_regex: for long 'indicator | country | date | value' files, must match exactly one indicator.
    """
    df = read_table(path)
    g = guess_columns(df)
    g.update(columns or {})
    if verbose:
        print(f"[{name}] columns used: {g}")
    if g.get("date") is None:
        raise ValueError(f"No date column found. Columns: {list(df.columns)}. Pass columns={{'date': ...}}")
    if kind == "auto":
        kind = "counts" if g.get("value") else "linelist"

    if kind == "counts":
        if g.get("value") is None:
            raise ValueError("No value column found; pass columns={'value': ...}")
        if g.get("indicator"):
            inds = sorted(df[g["indicator"]].dropna().astype(str).unique())
            if indicator_regex is None:
                raise ValueError("Several indicators in this file; pass indicator_regex. Available:\n  "
                                 + "\n  ".join(inds[:40]))
            mask = df[g["indicator"]].astype(str).str.contains(indicator_regex, case=False, regex=True)
            matched = sorted(df.loc[mask, g["indicator"]].astype(str).unique())
            if len(matched) != 1:
                raise ValueError(f"indicator_regex must match exactly ONE indicator; matched {matched}")
            if verbose:
                print(f"[{name}] indicator: {matched[0]}")
            df = df[mask]
        weekly = counts_to_weekly(df, g["date"], g.get("zone"), g["value"],
                                  cumulative=cumulative, verbose=verbose)
    elif kind == "linelist":
        if keep_status and not g.get("status") and verbose:
            print(f"[{name}] no status column found: keeping ALL cases")
        weekly = linelist_to_weekly(df, g["date"], g.get("zone"), g.get("status"), keep_status)
    else:
        raise ValueError("kind must be 'auto', 'counts' or 'linelist'")

    validate(weekly, name, expected_total)
    return weekly


def load_west_africa(path, **kw) -> pd.DataFrame:
    """West Africa 2014-16 (Zaire). Official total ~28,616 cases (confirmed+probable+suspected)."""
    kw.setdefault("expected_total", 28616)
    return load_outbreak(path, "WestAfrica_2014", **kw)


def load_uganda_2022(path, **kw) -> pd.DataFrame:
    """Uganda 2022 (Sudan virus), Global.health line list. ~142 confirmed / 164 incl. probable."""
    kw.setdefault("keep_status", ("confirmed",))
    kw.setdefault("expected_total", 142)
    return load_outbreak(path, "Uganda_2022", **kw)


# --------------------------------------------------------------------------
# Pre-training corpus
# --------------------------------------------------------------------------
def to_series(weekly: pd.DataFrame, outbreak: str, min_total: float = 10, min_weeks: int = 6) -> List[Dict]:
    """One entry per zone (+ '__total__'), trimmed to start at the first case."""
    out = []
    for z, g in weekly.groupby("zone"):
        g = g.sort_values("date")
        v = g["value"].to_numpy(float)
        nz = np.flatnonzero(v > 0)
        if nz.size == 0:
            continue
        v = v[nz[0]:]
        if v.sum() < min_total or len(v) < min_weeks:
            continue
        out.append(dict(outbreak=outbreak, zone=z, series=v, total=float(v.sum()),
                        start=g["date"].iloc[nz[0]]))
    return out


def normalise(v: np.ndarray, method: str = "log_peak") -> np.ndarray:
    """'log_peak': log1p then divide by the max; 'peak': divide by the max. Result in [0, 1]."""
    x = np.log1p(v) if method == "log_peak" else np.asarray(v, float)
    return x / (x.max() + 1e-8)


def make_windows(items: List[Dict], seq_len: int = 4, horizon: int = 1, method: str = "log_peak"):
    X, y, groups = [], [], []
    for it in items:
        v = normalise(it["series"], method)
        for i in range(seq_len, len(v) - horizon + 1):
            X.append(v[i - seq_len:i])
            y.append(v[i + horizon - 1])
            groups.append(it["outbreak"])
    if not X:
        raise ValueError("No windows produced - lower seq_len/min_weeks or check the data")
    return np.asarray(X, np.float32)[..., None], np.asarray(y, np.float32), np.asarray(groups)


def build_corpus(sources: Dict[str, pd.DataFrame], seq_len: int = 4, horizon: int = 1,
                 min_total: float = 10, method: str = "log_peak") -> Dict:
    """sources: {'WestAfrica_2014': weekly_df, 'Uganda_2022': weekly_df, ...}"""
    items: List[Dict] = []
    for name, weekly in sources.items():
        items += to_series(add_total(weekly), name, min_total=min_total, min_weeks=seq_len + horizon + 2)
    X, y, groups = make_windows(items, seq_len, horizon, method)
    print(f"corpus: {len(items)} series | {len(X)} windows | X{X.shape} | per outbreak: "
          f"{ {g: int((groups == g).sum()) for g in np.unique(groups)} }")
    return dict(series=items, X=X, y=y, groups=groups)


def leave_one_out(groups: np.ndarray, holdout: str):
    """Boolean masks (train, test) for leave-one-outbreak-out validation."""
    test = groups == holdout
    return ~test, test


# --------------------------------------------------------------------------
# CLI: download + show the columns so you can pick the right ones
# --------------------------------------------------------------------------
def download_and_inspect(data_dir="real_sources") -> List[Path]:
    """Download the West Africa (HDX) and Uganda (GitHub) files and print their columns.
    Call this directly from a notebook:  ro.download_and_inspect()"""
    root = Path(data_dir)
    files: List[Path] = []
    try:
        files += fetch_hdx(SOURCES["WestAfrica_2014"]["hdx_package"], root / "hdx")
    except Exception as e:  # network / API errors
        print(f"HDX download failed: {e}")
    try:
        files += fetch_github_files(SOURCES["Uganda_2022"]["github_repo"], root / "github")
    except Exception as e:
        print(f"GitHub download failed: {e}")
    for f in files:
        print(f"\n=== {f.name} ===")
        try:
            inspect(read_table(f))
        except Exception as e:
            print(f"  cannot read: {e}")
    return files


def main(argv=None) -> None:
    ap = argparse.ArgumentParser(description="Download real Ebola outbreak data and show its columns")
    ap.add_argument("--download", action="store_true")
    ap.add_argument("--data-dir", default="real_sources")
    a, _ = ap.parse_known_args(argv)  # ignores extra args injected by Jupyter/IPython
    if a.download:
        download_and_inspect(a.data_dir)


if __name__ == "__main__" and "ipykernel" not in __import__("sys").modules:
    main()
```
